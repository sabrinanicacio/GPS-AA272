```python
import numpy as np
import tifffile
from tqdm import tqdm
import matplotlib.pyplot as plt
from gpytoolbox import ray_triangle_intersect
import os
from mpl_toolkits.mplot3d.art3d import Poly3DCollection
import csv
import pandas as pd
```

```python
## Loading:

# Constants:
moon_radius = 1737.4  # km
earth_to_moon_dist = 384399 # km
num_gps_sats = 30
earth_ang = 0 # deg
gps_rad = 20180 #km
earth_rad = 6500 #km

# Image:
input_filename = "ldem_4.tif"
elev_km = tifffile.imread(input_filename)

# Image size:
img_width = elev_km.shape[1]
img_height = elev_km.shape[0]
print("Image Width =", img_width, "pixels, Image Height =", img_height, "pixels")

# Down sampling:
downsample_ratio = 1
downsample_elev = np.zeros((int(np.floor(img_height / downsample_ratio)), int(np.floor(img_width / downsample_ratio) )))
i0 = 0
j0 = 0
for i in range(img_width):
    if i % downsample_ratio == 0:
        for j in range(img_height):
            if j % downsample_ratio == 0:
                downsample_elev[j0][i0] = elev_km[j][i]
                j0 += 1
        j0 = 0
        i0 += 1


# Radius:
rad_map = downsample_elev + moon_radius
img_width = rad_map.shape[1]
img_height = rad_map.shape[0]
print("New Image Width =", img_width, "pixels, New Image Height =", img_height, "pixels")
print("Total pixel counts =", (img_width * img_height))
print(np.max(rad_map))


# Cutoff:
threshold_lat = -70
threshold_idx = 0
for j in range(img_height):
    lat = 90 - ((j+1) * 1.0 / (img_height+2)) * 180
    if lat < threshold_lat:
        threshold_idx = j 
        break

test_threshold_lat = -80
test_threshold_idx = 0
for j in range(img_height):
    lat = 90 - ((j+1) * 1.0 / (img_height+2)) * 180
    if lat < test_threshold_lat:
        test_threshold_idx = j 
        break

test_map = rad_map[test_threshold_idx:, :]
rad_map = rad_map[threshold_idx:,:]
img_width = rad_map.shape[1]
img_height = rad_map.shape[0]
print("With Latitude cutoff", threshold_lat, ", we cutoff at idx", threshold_idx, "and have new size:", img_width, ",", img_height)
print("With Latitude cutoff", test_threshold_lat, ", we cutoff the test points at idx", test_threshold_idx, "and have new size:", test_map.shape[1], ",", test_map.shape[0])
```

```python
## Confirm elevation map load is successful:
plt.imshow(rad_map, cmap="gray")
plt.colorbar()
plt.show()
print(rad_map.dtype, rad_map.shape)
print(rad_map.max())
print(rad_map.min())
print(rad_map[0][0])
```

```python
def get_lat_lon(i,j):
    lon = (i * 1.0 / img_width) * 360 - 180
    lat = threshold_lat - ((j+1) * 1.0 / (img_height+2)) * (threshold_lat + 90)
    return lat, lon

def get_moon_pos(elev, i, j):
    if j == 0:
        return np.array([0, 0, -1 * np.mean(elev[:,0])])
    if j == img_height + 1:
        return np.array([0, 0, np.mean(elev[:,img_height-1])])
    lat_r, lon_r = np.deg2rad(get_lat_lon(i,j))
    print(lat_r, ",", lon_r, "-->",np.cos(lat_r))
    x = elev[j-1][i] * np.cos(lat_r) * np.cos(lon_r)
    y = elev[j-1][i] * np.cos(lat_r) * np.sin(lon_r)
    z = elev[j-1][i] * np.sin(lat_r) 
    print(z)
    return np.array([x, y, z])

def get_ij(i):
    if i == img_width * img_height:
        return i 
    return (i % img_height, int(np.floor(i / img_height)))

def get_index(i,j):
    if j == 0:
        return img_width * img_height
    if j == img_height+1:
        return img_width * img_height + 1
    #print(i, ",", j, " --> ", int(i * img_width + (j-1)))
    return int(i * img_height + (j-1))

def get_testpoint_index(i,j):
    width = test_map.shape[1]
    height = test_map.shape[0]
    if j == height:
        return width * height
    #print(i, ",", j, " --> ", int(i * height + (j-1)))
    return int(i * height + (j-1))
```

```python
## Make Sphere points
points = np.zeros((img_height * img_width + 2, 3))
test_points = np.zeros((test_map.shape[1] * test_map.shape[0] + 1, 3))
test_map_height_dif = img_height - test_map.shape[0]
#print("Height Difference = ", test_map_height_dif)
for i in tqdm(range(img_width)):
    for j in range(img_height):
        print(j)
        pos = get_moon_pos(rad_map, i, j+1)
        #print(pos)
        points[get_index(i,j+1)][0] = pos[0]# np.vstack([points, ])
        points[get_index(i,j+1)][1] = pos[1]
        points[get_index(i,j+1)][2] = pos[2]
        if j + 1 > test_map_height_dif:
            test_points[get_testpoint_index(i,j+1 - test_map_height_dif)][0] = pos[0]
            test_points[get_testpoint_index(i,j+1 - test_map_height_dif)][1] = pos[1]
            test_points[get_testpoint_index(i,j+1 - test_map_height_dif)][2] = pos[2]
        #print(i, ", ", j, "-->", get_index(i,j+1))
        #print(get_moon_pos(rad_map, i, j))

# north_pole = get_moon_pos(rad_map, 0, 0)
# south_pole = get_moon_pos(rad_map, 0, img_height + 1)
points[img_height *img_width][2] = 1736.43123328
points[img_height *img_width + 1][2] = -1736.6809166
test_points[test_map.shape[1] * test_map.shape[0]][2] = -1736.6809166
```

```python
## Make Tri points
tris = np.zeros((img_width*img_height * 2, 3), dtype=int)
ind = 0
for i in tqdm(range(img_width)):
    for j in range(img_height):
        # Make triangle with this point in upper left:
        i1 = i
        j1 = j+1
        i2 = i
        j2 = j + 2
        i3 = (i+1) % img_width
        j3 = j + 1
        tris[ind][0] = get_index(i1,j1)
        tris[ind][1] = get_index(i2,j2)
        tris[ind][2] = get_index(i3,j3)
        ind += 1

        # Make triangle with this point in the lower right:
        i1 = i
        j1 = j + 1
        i2 = i
        j2 = j
        i3 = (i-1) % img_width
        j3 = j + 1
        tris[ind][0] = get_index(i1,j1)
        tris[ind][1] = get_index(i2,j2)
        tris[ind][2] = get_index(i3,j3)
        ind += 1

print(tris.shape)
```

```python
def get_horizon_curve(point, points, tris):
    num_angs = 30
    atol = 0.5 #deg
    phis = np.zeros(num_angs)
    thetas = np.zeros(num_angs)

    for i in tqdm(range(num_angs)):
        theta = 1.0 * i / num_angs * 360
        thetas[i] = theta
        s_theta = np.sin(np.deg2rad(theta))
        c_theta = np.cos(np.deg2rad(theta))
        phi = 0
        dphi = 45
        while dphi > atol:
            s_phi = np.sin(np.deg2rad(phi))
            c_phi = np.cos(np.deg2rad(phi))
            dir = np.array([ c_phi * s_theta, c_phi * c_theta, -1 * s_phi])
            sky = True
            for tri in tris:
                v0 = points[tri[0]]
                v1 = points[tri[1]]
                v2 = points[tri[2]]
                _, hits, _ = ray_triangle_intersect(point, dir, v0, v1, v2)
                if hits:
                    sky = False
                    break
            if sky:
                phi = phi - dphi
            else:
                phi = phi + dphi
            dphi = dphi / 2
            #print("Sky =", sky, " -->New phi = ", phi, "New dphi = ", dphi)
        phis[i] = phi

    return thetas, phis

south_pole_point = np.array([0,0, -1736.69])
crater_point = np.array([-7.75584093, 12.78123029, -1734.557])
thetas, phis = get_horizon_curve(crater_point, points, tris)
```

```python
def get_next_filename(base_name, ext):
    """
    Returns a filename like 'base_name.ext', 'base_name_1.ext', etc.,
    that doesn't already exist.
    """
    filename = f"{base_name}.{ext}"
    i = 1
    while os.path.exists(filename):
        filename = f"{base_name}_{i}.{ext}"
        i += 1
    return filename

def write_output_tiff():
    output_filename = get_next_filename("output", "tiff")
    input = tifffile.imread(input_filename)
    output = np.zeros((input.shape[0], input.shape[1], 3))
    minValue = np.min(input)
    maxValue = np.max(input)
    print(minValue)
    print(maxValue)
    print(input.shape[0])
    print(input.shape[1])
    for i in range(input.shape[1]):
        for j in range(input.shape[0]):
            index = get_index(int(np.floor(i / downsample_ratio)), int(np.floor(j / downsample_ratio) + 1))
            #print(int(np.floor(i / downsample_ratio)), ", ", (int(np.floor(j / downsample_ratio)) + 1), "-->", index)
            #print(index)
            num_visible = num_rays[index]
            rgb_val = (input[j][i] - minValue) / (maxValue - minValue)
            if num_visible == 0:
                output[j][i][0] = rgb_val
            elif num_visible == num_gps_sats:
                output[j][i][1] = rgb_val
            else:
                output[j][i][2] = rgb_val
    tifffile.imwrite(output_filename, output, photometric='rgb')
    return output_filename

output_filename = write_output_tiff()
img = tifffile.imread(output_filename)

# Display
plt.imshow(img)
plt.axis('off')  # optional, remove axes
plt.show()
```
